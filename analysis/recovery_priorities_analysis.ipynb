# Recovery Priorities Analysis
This notebook analyses anonymised or simulated community recovery priorities.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/recovery_priorities.csv')
df.head()
```

```python
priority_summary = df.groupby('priority_category')['priority_score'].mean().sort_values(ascending=False)
priority_summary
```

```python
priority_summary.plot(kind='barh', title='Average Recovery Priority Score')
plt.tight_layout()
plt.savefig('../visualisations/priority_scores.png')
plt.show()
```
