# Recovery Priorities Analysis

This notebook analyses a simulated dataset of community recovery priorities across three stakeholder groups.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/survey_responses.csv')
df.head()
```

```python
summary = df.groupby(['stakeholder_group', 'priority_category'])['priority_score'].mean().unstack()
summary.round(2)
```

```python
ax = summary.plot(kind='bar', figsize=(10, 5))
ax.set_title('Average recovery priority by stakeholder group')
ax.set_ylabel('Average priority score')
ax.set_xlabel('')
plt.tight_layout()
plt.savefig('../visualisations/priorities_by_group.svg')
plt.show()
```

```python
normalised = summary.div(summary.sum(axis=1), axis=0)

fig, ax = plt.subplots(figsize=(9, 4))
im = ax.imshow(normalised.values, cmap='YlGnBu', aspect='auto')
ax.set_xticks(range(len(normalised.columns)))
ax.set_xticklabels(normalised.columns, rotation=25, ha='right')
ax.set_yticks(range(len(normalised.index)))
ax.set_yticklabels(normalised.index)

for i in range(normalised.shape[0]):
    for j in range(normalised.shape[1]):
        ax.text(j, i, f'{normalised.iloc[i, j]:.0%}', ha='center', va='center')

ax.set_title('Share of recovery priorities by stakeholder group')
fig.colorbar(im, ax=ax, label='Share of priorities')
plt.tight_layout()
plt.savefig('../visualisations/priority_heatmap.svg')
plt.show()
```
